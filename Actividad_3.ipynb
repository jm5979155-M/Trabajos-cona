```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re
```

```python
url = "https://elfondoenlinea.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
```

```python
libros = []

contenedores = soup.find_all('div', class_=lambda x: x and 'item-slick' in x.split())

print(f"Se encontraron {len(contenedores)} libros")
```

```python
for contenedor in contenedores:
    titulo = contenedor.find('p', class_='titulo-mini-slider')
    titulo = titulo.text.strip() if titulo else ''

    autor = contenedor.find('p', class_='autor-mini-slider')
    autor = autor.text.strip() if autor else ''

    precio = contenedor.find('ul', class_='nav fce-precios')
    precio = ' '.join(precio.text.strip().split()) if precio else ''

    libros.append({'Título': titulo, 'Autor': autor, 'Precio': precio})
```

```python
df = pd.DataFrame(libros)
print("="*50)
print("DATOS EXTRAÍDOS CORRECTAMENTE")
print("="*50)
print(f"Total de libros: {len(df)}")
print("\nPrimeros 10 libros:")
print(df.head(10))
```
