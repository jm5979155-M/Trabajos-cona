# Traductor básico Español → Inglés

## Descripción
Este programa utiliza un diccionario en Python para traducir palabras del español al inglés.  
Si la palabra no se encuentra en el diccionario, devuelve un mensaje indicando que no fue encontrada.

## Funcionamiento
- Se define un diccionario con palabras en español y su traducción.
- Se crea una función que busca la palabra en el diccionario.
- Se imprime el resultado de la traducción.

## Nota
Las palabras deben escribirse exactamente igual a como están en el diccionario.

```python
dic = {"hola": "hello", "mundo": "world", "carro": "car", "casa": "home", "piso": "floor", "mesa": "table", "lapiz": "pencil", "lentes": "glasses", "silla": "chair", "teclado": "keyboard"}

def translate(word):
  return dic.get(word, "no encontrada")

word = "casa"
print(translate(word))
```
