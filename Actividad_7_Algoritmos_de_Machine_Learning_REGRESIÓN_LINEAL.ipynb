Ahora separaremos las características de la etiqueta, lo pondremos en la variable x y y

```python
X = datos_na.drop(["median_house_value"], axis=1)
y = datos_na["median_house_value"]
```

Ahora vamos a separar los datos en dos partes. Conjunto de entrenamiento y uno de pruebas

```python
from sklearn.model_selection import train_test_split

X_ent, X_pru, y_ent, y_pru = train_test_split(X , y, test_size=.2)
```

Ahora inspeccionamos

```python
X_pru.shape
```

Ahora crearemos el modelo

```python
from sklearn.linear_model import LinearRegression

modelo=LinearRegression()
```

Ahora se va a entrenar

```python
modelo.fit(X_ent, y_ent)
```

Ahora haremos predicciones

```python
predicciones = modelo.predict(X_pru)
```

Aquí haremos una variable que sea un diccionario para mostrar que fue lo que predijo y elvalor real

```python
comparativa = {"Prediccion": predicciones, "Valor Real": y_pru}
pd.DataFrame(comparativa)
```

Vamos a ver algo llamado "Overfitting o sobreajuste" que es cuando el modelo se aprende demasiado los datos que pued llegar a fallar

```python
print(modelo.score(X_ent, y_ent))
print(modelo.score(X_pru, y_pru))
```

Podemos calcular el error (agrupamos todos los datos y vemos si estamos fallando en algo) con lo siguiente

```python
from sklearn.metrics import mean_squared_error
import numpy as np

mse = mean_squared_error(y_pru, predicciones)
```

Imprimimos el resultado

```python
print(mse)
```

En esta parte obtendremos la raiz cuadrada

```python
rmse =np.sqrt(mse)
```

```python
print(rmse)
```
