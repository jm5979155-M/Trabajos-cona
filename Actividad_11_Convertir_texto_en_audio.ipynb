```python
!pip install gtts

from gtts import gTTS
from IPython.display import Audio

# Escribimos el texto a convertir
texto = "Buenos dias profe este es mi trabajo de convertir texto a audio"

# Creamos el audio
tts = gTTS(text=texto, lang='es')

# Guardamos el archivo
tts.save("audio.mp3")

print("Audio generado correctamente.")

# Esta linea es la que reproduce el audio
Audio("audio.mp3")
```
